```python
#5th question
# Descriptive statistics
print("Descriptive Statistics:\n")
print(df.describe())

# Mean values grouped by species
species_mean = df.groupby("Species").mean()

print("\nAverage Dimensions of Each Species:\n")
print(species_mean)

# Find species with overall largest dimensions
species_mean["overall_size"] = species_mean.mean(axis=1)

largest_species = species_mean["overall_size"].idxmax()

print("\nSpecies with Largest Overall Dimensions:")
print(largest_species)
```
