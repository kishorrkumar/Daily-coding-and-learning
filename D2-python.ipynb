```python
year = int(input("enter year: "))

if year%400== 0:
    print(year, "is a leap year")
elif year%100 == 0:
    print(year, "is not a leap year")
elif year%4 == 0:
    print(year, "is a leap year")

else:
    print(year, "is not a leap year")
```

```python
n1 = int(input("enter num1: "))
n2 = int(input("enter num2: "))
operator =input("enter operator: ")

if operator == "+":
    print("Result: ", n1 + n2)
elif operator == "-":
    print("Result: ", n1 - n2)
elif operator == "*":
    print("Result: ", n1 * n2)
elif operator == "/":
    if n2 != 0:
        print("Result: ", n1/n2)
    else:
        print("cant divisible by zero!")
```
