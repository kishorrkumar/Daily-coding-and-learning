```python
def missing_summary(titanic):
    missing = titanic.isnull().sum()

    percent = (missing / len(titanic)) * 100

    result = pd.DataFrame({
        'missing count': missing,
        'missing %': percent.round(2)
    })

    return result[result['missing count'] > 0].sort_values(
        by='missing %',
        ascending=False
    )
```

```python
missing_summary(titanic)
```
