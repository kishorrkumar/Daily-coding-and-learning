```python
#Pyramid
n = 10 
for i in range(n):

    for j in range(i,n):
        print(" ", end = " ")
    for j in range(i):
        print("*",end = " ")
    for j in range(i+1):
        print("*",end = " ")
    print()
```

```python
#Inverse Pyramid
n = 10

for i in range(n):

    for j in range(i):
        print(" ", end = " ")
    for j in range(i,n):
        print("*", end = " ")
    for j in range(i + 1,n):
        print("*", end = " ")
        
    

    print()
```

```python
#Diamond
n = 10 
for i in range(n):

    for j in range(i,n):
        print(" ", end = " ")
    for j in range(i):
        print("*",end = " ")
    for j in range(i+1):
        print("*",end = " ")
    print()
    
for i in range(n):
    for j in range(i+1):
        print(" ", end = " ")
    for j in range(i,n):
        print("*",end = " ")
    for j in range(i+1,n):
        print("*",end = " ")

    print()
```

```python
#Sunglasses
n = 5
for i in range(n):
    for j in range(i):
        print(" ", end = " ")
    for j in range(i,n):
        print("*", end = " ")
    for j in range(i+1,n):
        print("*", end = " ")
    print()
a = 5
for i in range(a):
    for j in range(i,a-1):
        print(" ", end = " ")
    for j in range(i):
        print("*", end = " ")
    for j in range(i+1):
        print("*", end = " ")
    print()
```

```python
#Double hill
n = 5
for i in range(n):
    for j in range(i,n):
        print( " ", end = " " )
    for j in range(i):
        print("*", end = " " )
    for j in range(i+1):
        print("*", end = " ")
    for j in range(i+1,n):
        print( " ", end = " " )
    for j in range(i+1,n):
        print( " ", end = " " )
    for j in range(i):
        print("*", end = " " )
    for j in range(i+1):
        print("*", end = " ")

    print()
    
```
