```python
#list methods
fruits = ["apple", "banana", "Kiwi", "guava"]

fruits.append("orange")
print(fruits)

fruits.insert(1, "grapes")
print(fruits)


fruits.sort()
print(fruits)

print(fruits[2:5])

print(fruits[:5])
print(fruits[5:])
print(fruits[::2])
print(fruits[::-1])
```

```python
#List comprehension
#expression------->iterate------>condition
[x ** 2 for x in range(11)]
```

```python
#print even number
list  =  []

for i in range(10):
    if i%2==0:
        list.append(i)
print(list)
```

```python
#print even numbers only
even_num = [num for num in range(10) if num%2== 0 ]

print(even_num)
```

```python
#Pair 
list1 = [1, 2, 3, 4, 5]
list2 = ["a", "b", "c","d"]
[ [i,j] for i in list1 for j in list2 ]
```

```python
#length of each word
words = ["hello", "world", "python", "list"]

count = [len(i) for i in words]
print(count)
```
