```python
import pandas as pd

def missing_summary(df):
    missing = df.isnull().sum()

    percent = (missing / len(df)) * 100

    result = pd.DataFrame({
        'missing count': missing,
        'missing %': percent.round(2)
    })

    return result[result['missing count'] > 0].sort_values(
        by='missing %',
        ascending=False
    )
```

```python
missing_summary(df)
```
